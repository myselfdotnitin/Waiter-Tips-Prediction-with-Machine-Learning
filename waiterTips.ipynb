## Waiter Tips Prediction

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

data = pd.read_csv("tips.csv")
print(data.head())
```

lets move forward by ananlysing all the factors affecting waiter tips:
1. Total bill paid
2. Number of people at a table
3. The day of the week

```python
figure = px.scatter(data_frame = data, x="total_bill",
                    y="tip", size="size", color="day", trendline="ols")
figure.show()
```

```python
figure = px.scatter(data_frame=data, x="total_bill",
                    y="tip", color="sex", trendline="ols")
figure.show()
```

```python
figure = px.scatter(data_frame=data, x="total_bill",
                    y="tip", size="size", color="time", trendline="ols")
figure.show()
```
